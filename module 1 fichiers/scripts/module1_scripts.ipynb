### Trier les fichiers par extension

Ce qu'on va faire : pathlib va lire chaque fichier, 

détecter son extension, et shutil va le déplacer dans le bon sous-dossier.

```python
# --- pathlib pointe vers ton dossier en désordre ---
source = Path(r"C:\Users\HP\Downloads\test_entreprise")

# --- le dossier de destination ---
destination_base = Path(r"C:\Users\HP\Downloads\test_entreprise")

# --- la règle : quelle extension va dans quel sous-dossier ---
categories = {
    ".csv": "csv",
    ".xlsx": "excel",
    ".json": "json",
    ".txt": "texte"
}

# --- pathlib liste les fichiers, shutil les déplace ---
for fichier in source.iterdir():
    if fichier.is_file():                              # on ignore les sous-dossiers
        extension = fichier.suffix.lower()             # .CSV devient .csv

        if extension in categories:
            dossier_cible = destination_base / categories[extension]
            dossier_cible.mkdir(parents=True, exist_ok=True)   # crée le sous-dossier si besoin
            shutil.move(fichier, dossier_cible / fichier.name) # déplace le fichier
            print(f"Déplacé : {fichier.name} → {categories[extension]}/")
            print("-" * 40)
        else:
            print(f"Extension non reconnue : {fichier.name}")
```

### Renommer les fichiers en snake_case

Ce qu'on va faire : pathlib lit le nom de chaque fichier, 

on le transforme en snake_case minuscule,

puis os applique le renommage physiquement.

```python
# les dossiers à traiter 

dossiers = [
    Path(r"C:\Users\HP\Downloads\test_entreprise\csv"),
    Path(r"C:\Users\HP\Downloads\test_entreprise\excel"),
    Path(r"C:\Users\HP\Downloads\test_entreprise\json"),
    Path(r"C:\Users\HP\Downloads\test_entreprise\texte")
]

for dossier in dossiers:
    for fichier in dossier.iterdir():
        if fichier.is_file():

            #  transformation du nom 
            nouveau_nom = fichier.stem.lower()       # met en minuscules
            nouveau_nom = nouveau_nom.replace(" ", "_")   # espaces → underscores
            nouveau_nom = nouveau_nom.replace("-", "_")   # tirets → underscores
            nouveau_nom = f"{nouveau_nom}{fichier.suffix.lower()}"  # recolle l'extension

            nouveau_chemin = fichier.parent / nouveau_nom  # reconstruit le chemin complet du fichier avec son nouveau nom, 
                                                           #dans le même dossier qu'avant

            #  renommage uniquement si le nom du fichier change 
            if fichier.name != nouveau_nom:
                os.rename(fichier, nouveau_chemin)
                print(f"Renommé : {fichier.name} → {nouveau_nom}")
            else:
                print(f"Inchangé : {fichier.name}")
```

---
### Valider les fichiers avant de les traiter

Ce qu'on va faire : pathlib va inspecter chaque fichier CSV avant de le lire.

Si un fichier est vide, on l'identifie et on l'écarte. Si il est valide, pandas le lit normalement.

```python

dossier_csv = Path(r"C:\Users\HP\Downloads\test_entreprise\csv")

for fichier in dossier_csv.glob("*.csv"):

    # pathlib vérifie la taille du fichier en octets 
    taille = fichier.stat().st_size

    if taille == 0:
        print(f"ATTENTION fichier vide : {fichier.name}")
        print("-" * 40)
    else:
        df = pd.read_csv(fichier)
        print(f"OK : {fichier.name} — {len(df)} lignes")
        print("-" * 40)
```
